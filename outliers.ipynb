```python
import numpy as np
```

```python
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
# find the quartile
Q1 = np.percentile(arr, 25)
Q2 = np.percentile(arr, 50)
Q3 = np.percentile(arr, 75)
```

```python
print(Q1)
```

```python
print(Q2)
```

```python
print(Q3)
```

```python
# find the IQR
IQR = Q3 - Q1
```

```python
print(IQR)
```

```python
# find the UPPER FENCE and LOWER FENCE
UF = Q3 + (1.5*IQR)
LF = Q1 - (1.5*IQR)
```

```python
print(UF)
```

```python
print(LF)
```

```python
# for printing the array without the outlier 
l = []
for i in arr:
    if i >= LF and i and i <= UF:
        l.append(i)

l
# convert the list into the array agian
arr2 = np.array(l)
```

```python
# with outlier
print(arr)
# without outlier
print(arr2)
```
